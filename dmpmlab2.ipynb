```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 15, 20, 25, 30]

fig, axs = plt.subplots(2, 3, figsize=(15, 8))

# Line Chart
axs[0, 0].plot(x, y, marker='o')
axs[0, 0].set_title("Line Chart")

# Bar Plot
axs[0, 1].bar(x, y, color='green')
axs[0, 1].set_title("Bar Plot")

# Scatter Plot
axs[0, 2].scatter(x, y, color='red')
axs[0, 2].set_title("Scatter Plot")

# Histogram
axs[1, 0].hist(y, bins=5, color='orange', edgecolor='black')
axs[1, 0].set_title("Histogram")

# Pie Chart
axs[1, 1].pie(y, labels=x, autopct='%1.1f%%')
axs[1, 1].set_title("Pie Chart")

# Box Plot
axs[1, 2].boxplot(y)
axs[1, 2].set_title("Box Plot")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
sales = [120, 150, 160, 170, 200]

plt.plot(months, sales, marker='o', color='blue')
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

subjects = ['Math', 'Science', 'English', 'Computer']
marks = [60, 70, 80, 90]

plt.bar(subjects, marks, color='green')
plt.title("Student Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.show()
```

```python
import matplotlib.pyplot as plt

height = [150, 155, 160, 165, 170, 175, 180]
weight = [50, 55, 60, 65, 68, 72, 78]

plt.scatter(height, weight, color='red')
plt.title("Height vs Weight")
plt.xlabel("Height (cm)")
plt.ylabel("Weight (kg)")
plt.show()
```

```python
import matplotlib.pyplot as plt

languages = ['Python', 'Java', 'C++', 'JavaScript']
students = [40, 25, 20, 15]

plt.pie(students,
        labels=languages,
        autopct='%1.1f%%',
        startangle=90)

plt.title("Programming Language Preference")
plt.show()
```
